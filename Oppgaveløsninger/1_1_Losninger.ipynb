## Oppgave 3
**Kommentar:** Vi starter med 20 kr, og jobber oss nedover til 1 kr.

```python
178 // 20           # Gir 8 hele 20-kr
178 % 20            # Da er det 18 kr igjen
18 // 10            # Gir 1 hel 10-kr
18 % 10             # Da er det 8 kr igjen
8 // 5              # Gir 1 hel 5-kr
8 % 5               # Da er det 3 kr igjen
3 // 1              # Gir 3 hele 1-kr
3 % 1               # Da er det 0 kr igjen
```

**Kommentar:** Fordelingen er altså 8 stk 20-kr, 1 stk 10-kr, 1 stk 5-kr og 3 stk 1-kr.
