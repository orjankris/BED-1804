## Oppgave 1 a)

```python
f = lambda x: x ** 2 - 4
print(f'f(-3) = {f(-3)}, f(0) = {f(0)}, f(5) = {f(5)}')
```

## Oppgave 1 b)

```python
g = lambda x_liste: [x ** 2 - 4 for x in x_liste]
x = list(range(-5, 6))

print(f'x-verdier: {x} \ny-verdier: {g(x)}')
```

## Oppgave 2 a)

```python
butikker = [['Sydnes', 11, 178], ['Østnes', 24, -304], ['Vestnes', 5, 69], ['Nordnes', 17, -125]]

sortert_ansatte = sorted(butikker, key = lambda x: x[1])
sortert_budsjett = sorted(butikker, key = lambda x: abs(x[2])) 

print('Butikkene sortert etter antall ansatte:')
print(sortert_ansatte)
print('Butikkene sortert etter budsjettavvik:')
print(sortert_budsjett)
```

## Oppgave 2 b)

```python
rpa_liste = list(map(lambda x: x[2] / x[1], butikker))

for indeks, rpa in enumerate(rpa_liste):
    print(f'{butikker[indeks][0]}: {rpa:.2f}')
```

**Kommentar:** I moderne Python er dette vanligere enn linje 1:

```python
rpa_liste = [x[2] / x[1] for x in butikker] 
```

## Oppgave 3

```python
def finn_vinner(aksjer):
    vinner = max(aksjer, key = lambda x: x[2] / x[1])
    return vinner[0]

portefolje = [['Equinor', 250, 300], ['DNB Bank', 180, 240], ['Frontline', 280, 350]]
finn_vinner(portefolje)
```
