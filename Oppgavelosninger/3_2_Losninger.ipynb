## Oppgave 1 a)
**Kommentar:** Ikke lag en variabel som heter `sum`, da dette er en funksjon i Python.

```python
# Vær obs på å ikke lage en variabel som heter sum!
slutt = 1000
sum_alle = 0

for tall in range(slutt + 1):
    sum_alle += tall

print(f'Summen av tallene mellom 0 og {slutt} blir {sum_alle}')
```

## Oppgave 1 b)

```python
slutt = 1000
sum_par, sum_odde = 0, 0

for tall in range(slutt + 1):
  if tall % 2 == 0:
    sum_par += tall
  else:
    sum_odde += tall

print(f'Summen av partallene mellom 0 og {slutt} blir {sum_par}')
print(f'Summen av oddetallene mellom 0 og {slutt} blir {sum_odde}')
```

## Oppgave 2 a)

```python
innskudd = [8000, 10000, 11000, 15000, 16000]
sparetid = len(innskudd)
rente = 0.05
saldo = 0

for aar, belop in enumerate(innskudd):
    saldo += belop * (1 + rente) ** (sparetid - aar)

print(f'Saldo etter {sparetid} år er {saldo:.2f} kr')
```

## Oppgave 2 b)

```python
innskudd_etter_rente = [belop * (1 + rente) ** (sparetid - aar) for aar, belop in enumerate(innskudd)]

print(f'Saldo etter {sparetid} år er {sum(innskudd_etter_rente):.2f} kr')
```

## Oppgave 3

```python
kurs_1 = ['Hans', 'Mons', 'Kari', 'Jens', 'Mari', 'Guri', 'Nils']
kurs_2 = ['Guri', 'Siri', 'Nils', 'Sara', 'Mons', 'Lara', 'Kari']
begge_kurs = []

for deltaker_1 in kurs_1:
    for deltaker_2 in kurs_2:
        if deltaker_1 == deltaker_2:
            begge_kurs.append(deltaker_1)

print('Følgende ansatte har deltatt på begge kursene:')
print(begge_kurs)
```

**Kommentar:** Dette er en enklere variant hvis det ikke eksplisitt ble bedt om to for-løkker:

```python
begge_kurs = []

for deltaker in kurs_1:
    if deltaker in kurs_2:
        begge_kurs.append(deltaker)

print('Følgende ansatte har deltatt på begge kursene:')
print(begge_kurs)
```
