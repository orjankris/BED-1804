## Oppgave 1 a)

```python
import sympy as sp

x = sp.symbols('x')
P = -x**2 + 10*x - 9
sp.plot(P, (x, 0, 10))
```

## Oppgave 1 b)

```python
x_null_profitt = sp.solve(P)
x_maks_profitt = sp.solve(sp.diff(P))[0]
P_maks_profitt = P.subs(x, x_maks_profitt)

print(f'Antall enheter for null profitt er {x_null_profitt}.')
print(f'Antall enheter for maks profitt er {x_maks_profitt}.')
print(f'Den maksimale profitten er {P_maks_profitt} tusen kr.')
```

## Oppgave 2 a)

```python
K = x**3 - 6*x**2 + 15*x + 10
K_der = sp.diff(K)
K_der_4 = K_der.subs(x, 4)
x_vendepunkt = sp.solve(sp.diff(K, x, 2))[0]
K_vendepunkt = K.subs(x, x_vendepunkt)

print(f"K'(x) = {K_der}, og K'(4) = {K_der_4}")
print(f'Vendepunktet er ({x_vendepunkt}, {K_vendepunkt})')
```

## Oppgave 2 b)

```python
snitt_kost = (1/5) * sp.integrate(K, (x, 0, 5))
print(f'Gjennomsnittsverdien er {snitt_kost:.2f}')
```

## Oppgave 3

```python
x, y, l = sp.symbols('x, y, l', positive=True)

U = x**0.4 * y**0.6
L = U - l * (2*x + y - 50)

likning_1 = sp.Eq(sp.diff(L, x), 0)
likning_2 = sp.Eq(sp.diff(L, y), 0)
likning_3 = sp.Eq(sp.diff(L, l), 0)

losning = sp.solve([likning_1, likning_2, likning_3], dict=True)
x_maks = losning[0][x]
y_maks = losning[0][y]
U_maks = U.subs({x: x_maks, y: y_maks})

print(f'Maksimal nytte oppnås ved å bruke {x_maks:.0f} av A og {y_maks:.0f} av B.')
print(f'Den maksimale nytten er {U_maks:.2f}.')
```
