## Oppgave 1 a)
**Kommentar:** `[-1]` gir siste element, `[-2]` gir nest siste element, og så videre.

```python
aksjer = ['EQNR', 'DNB', 'STB', 'NHY', 'YAR', 'TEL', 'KOG']
print(aksjer[0])
print(aksjer[6], aksjer[-1])
```

## Oppgave 1 b)
**Kommentar:** `[:N]` gir elementene fra starten til, men ikke med, indeks N.  
`[N:]` gir elementene fra og med indeks N til slutten.

```python
print(aksjer[0:3], aksjer[:3])
print(aksjer[4:7], aksjer[4:])
```

## Oppgave 2 a)
**Kommentar:** Bruk `.append()` for å legge til element bakerst, ellers bruk `.insert()`.

```python
handleliste = ['Cola', 'Solo', 'Epler', 'Druer', 'Rømme', 'Kefir']
handleliste.append('Fløte')
print(handleliste)
handleliste.insert(4, 'Bananer')
print(handleliste)
```

## Oppgave 2 b)
**Kommentar:** Bruk alternativt `del` eller `.pop()` for å fjerne element på gitt indeks.

```python
handleliste.remove('Epler')
print(handleliste)
handleliste[1] = 'Fanta'
print(handleliste)
```
