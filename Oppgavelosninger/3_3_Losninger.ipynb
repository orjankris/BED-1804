## Oppgave 1 a)

```python
budsjett = 500
brukt = 0

while brukt < budsjett:
    pris = float(input('Legg inn pris på is: '))
    brukt += pris

print(f'Budsjettet på {budsjett} kr er brukt opp.')
```

## Oppgave 1 b)

```python
budsjett = 500
brukt, antall_tot, antall_30 = 0, 0, 0

while brukt < budsjett:
    pris = float(input('Legg inn pris på is: '))
    brukt += pris
    antall_tot += 1
    if pris > 30:
        antall_30 += 1

print(f'Budsjettet på {budsjett} kr er brukt opp.')
print(f'Du har kjøpt {antall_tot} iser, der {antall_30} kostet mer enn 30 kr.')
```

## Oppgave 2

```python
saldo = 100000
rente = 0.05
uttak = 10000
aar = 0

while saldo * (1 + rente) >= uttak:
    aar += 1
    saldo = saldo * (1 + rente) - uttak
    print(f'{f'Saldo etter år {aar}:':20}{saldo:10.2f}')
```

## Oppgave 3

```python
kode_riktig = '1234'
forsok_maks = 3
forsok = 1
kode_bruker = input(f'Forsøk {forsok}: Legg inn 4-sifret kode: ')

while forsok < forsok_maks and kode_bruker != kode_riktig:
      print('Koden er feil.')
      forsok += 1
      kode_bruker = input(f'Forsøk {forsok}: Legg inn 4-sifret kode: ')

if kode_bruker == kode_riktig:
      print('Koden er riktig.')
else:
      print('Koden er feil. Du har ikke flere forsøk.')
```
