## Oppgave 1 a)

```python
def baklengs(ordet):
    antall = len(ordet)
    ord_baklengs = ordet[::-1]
    print(f'Ordet {ordet} består av {antall} tegn, og er baklengs gitt som {ord_baklengs}')

baklengs('Økonomi')
```

## Oppgave 1 b)

```python
def ramme(ordet):
    lengde = len(ordet)
    print('┌' + '─' * (lengde + 2) + '┐')
    print('│ ' + ordet + ' │')
    print('└' + '─' * (lengde + 2) + '┘')

ramme('Python-lab')
```

## Oppgave 2 a)

```python
def sluttsaldo(sparebelop, antall_aar, rente):
    saldo = sparebelop * (1 + rente) ** antall_aar
    return round(saldo, 2)

sluttsaldo(10000, 10, 0.1)
```

## Oppgave 2 b)

```python
def spareslutt(sparebelop, rente):
    antall_aar = len(sparebelop)
    saldo = 0
    for aar, belop in enumerate(sparebelop):
        saldo += belop * (1 + rente) ** (antall_aar - aar)
    return round(saldo, 2)

spareslutt([7000, 11000, 9000, 5000, 14000], 0.05)

# Kommentar: Linje 3 - 6 kan erstattes med:
#   belop_etter_rente = [belop * (1 + rente) ** (antall_aar - aar) for aar, belop in enumerate(sparebelop)]
#   return(round(sum(belop_etter_rente), 2))
```

## Oppgave 3

```python
def serielaan(laanebelop, lopetid, rentesats):
    avdrag = laanebelop / lopetid
    terminbelop = []
    for aar in range(lopetid):
        restlaan = laanebelop - aar * avdrag
        rente = restlaan * rentesats
        terminbelop.append(round(avdrag + rente, 2))
    return terminbelop

serielaan(300000, 5, 0.07)
```
