## Oppgave 1 a)

```python
import sympy as sp

uttrykk_1 = (sp.Rational(1, 2) + 1 - sp.Rational(1, 5)) / (sp.Rational(1, 3) + sp.Rational(3, 4))
display(uttrykk_1)

uttrykk_2 = (3*sp.sqrt(12)) / (2*sp.sqrt(27) - sp.sqrt(3))
display(uttrykk_2)

uttrykk_3 = (2**sp.Rational(1, 2) * 4**sp.Rational(1, 4)) / 16**sp.Rational(1, 8)
display( uttrykk_3)
```

## Oppgave 1 b)

```python
x = sp.symbols('x')

uttrykk = (x**18 - 1) / (x**6 - x**3 + 1)
display(sp.simplify(uttrykk))

uttrykk_2 = sp.sqrt(x+1)**5 - x * (x + 2) * sp.sqrt(x + 1)
display(sp.simplify(uttrykk_2))

uttrykk_3 = x**3 + 3*x**2 - 4*x - 12
display(sp.factor(uttrykk_3))
```

## Oppgave 2 a)

```python
likning_1 = sp.Eq(x**3 - 4*x, 3*x**2 -12)
display(sp.solve(likning_1))

likning_2 = sp.Eq((x + 1) / (x - 2), 3)
display(sp.solve(likning_2))

likning_3 = sp.Eq(5**(2*x + 1), 125**(x - 1))
display(sp.solve(likning_3))
```

## Oppgave 2 b)

```python
ulikhet_1 = sp.Ge(x**2 - 6, 5*x)
display(sp.solve(ulikhet_1))

ulikhet_2 = sp.Lt(x**2 + 10, 5*x)
display(sp.solve(ulikhet_2))

ulikhet_3 = sp.Gt((x + 1)/(x - 2), 3)
display(sp.solve(ulikhet_3))
```

## Oppgave 3
**Kommentar:** `losning` er en liste der hvert element representerer én løsning i form av en ordbok.  
Utelates `positive=True` blir det tre løsninger.

```python
p, q, a = sp.symbols('p, q, a', positive=True)

likning_1 = sp.Eq(p**2 - 100, 2*a - q)
likning_2 = sp.Eq(q - a, 14 + 3*p)
likning_3 = sp.Eq(p*q, 320)

losning = sp.solve((likning_1, likning_2, likning_3), dict=True)

print('Likningssystemet har følgende løsning:')
print(f'Pris = {losning[0][p]} \nSolgte varer = {losning[0][q]}')
```
