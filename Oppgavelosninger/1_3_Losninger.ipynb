## Oppgave 2 a)
**Kommentar:** I `print()` skiller man oftest variabler med komma, som konverterer til streng og gir mellomrom.  
Men siden det ikke skal være mellomrom før utropstegnet, så adderer vi de to strengene.

```python
vare = input('Legg inn en vare: ')
print(vare.capitalize(), 'er nå på tilbud, løp og kjøp', vare.lower() + '!')
```

## Oppgave 2 b)
**Kommentar:** Det fungerer helt fint å la tilbudsprisen være av datatypen streng her.

```python
vare, pris = input('Legg inn en vare og dens tilbudspris (skill med komma, uten mellomrom): ').split(',')
print(vare.capitalize(), 'er nå på tilbud, løp og kjøp', vare.lower(), 'til bare', pris, 'kr!')
```

## Oppgave 3 a) 

```python
tall = float(input('Legg inn et tilfeldig tall: '))
print(f'Tallet er {tall:.2f}, eller {tall:.2%}, eller {tall:.2e}')
```

## Oppgave 3 b)
**Kommentar:** Når man angir bredde blir strenger venstrejustert og tall høyrejustert som standard.  
I koden nedenfor er `<` og `>` satt på, men det er altså ikke nødvendig.

```python
print(f'{'Desimalform:':<20}{tall:>10.2f}')
print(f'{'Prosentform:':<20}{tall:>10.2%}')
print(f'{'Standardform:':<20}{tall:>10.2e}')
```
