## Oppgave 1 a)

```python
import matplotlib.pyplot as plt

resultater = [1253, 1571, 1694, 1438, 1421]
aar = [2021, 2022, 2023, 2024, 2025]

plt.bar(aar, resultater)
plt.title('Resultater for Pythonius AS')
plt.xlabel('År')
plt.ylabel('MNOK')
plt.show()
```

## Oppgave 1 b)

```python
import numpy as np

snitt = np.mean(resultater)

plt.bar(aar, resultater, label='Årlige resultater')
plt.title('Resultater for Pythonius AS')
plt.xlabel('År')
plt.ylabel('MNOK')
plt.axhline(snitt, color='red', label='Gjennomsnitt')
plt.legend()
plt.show()
```

## Oppgave 2 a)

```python
kurser = [312.92, 324.17, 323.04, 329.38, 340.11, 352.83, 337.73, 334.48, 339.04, 336.25]
dager = np.arange(32, 42)

fig, ax = plt.subplots()
ax.plot(dager, kurser, zorder=0)
ax.scatter(dager, kurser, color='red')
ax.set_xticks(dager)
plt.show()
```

## Oppgave 2 b)

```python
stigningstall, konstant = np.polyfit(dager, kurser, 1)
trendlinje = stigningstall * dager + konstant

ax.plot(dager, trendlinje, color='green', label='Trendlinje')
ax.legend()
fig
```

## Oppgave 3

```python
antall_rod = [1232, 1411, 1195]
antall_bla = [1681, 1593, 1468]
antall_gronn = [957, 1168, 901]
kvartaler = ['1. kvartal', '2. kvartal', '3. kvartal']

fig, (ax1, ax2, ax3) = plt.subplots(3, 1)
fig.suptitle('Kvartalsvise arbeidstimer for avdelingene rød, blå og grønn')

ax1.set_facecolor('red')
ax1.barh(kvartaler, antall_rod, color='black')
ax1.set_xlim([0, 1800])
ax2.set_facecolor('blue')
ax2.barh(kvartaler, antall_bla, color='black')
ax2.set_xlim([0, 1800])
ax3.set_facecolor('green')
ax3.barh(kvartaler, antall_gronn, color='black')
ax3.set_xlim([0, 1800])

plt.tight_layout()
plt.show()
```
