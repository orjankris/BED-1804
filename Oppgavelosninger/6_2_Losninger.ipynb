## Oppgave 1 a)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-1, 3, 100)
f_0 = x**3 - 3*x**2 + 2
f_1 = 3*x**2 - 6*x
f_2 = 6*x - 6

fig, ax = plt.subplots(3, 1)
ax[0].plot(x, f_0)
ax[0].set_title('Den originale funksjonen')
ax[1].plot(x, f_1)
ax[1].set_title('Den deriverte av funksjonen')
ax[2].plot(x, f_2)
ax[2].set_title('Den dobbeltderiverte av funksjonen')
plt.tight_layout()
plt.show()
```

## Oppgave 1 b)

```python
for i in range(3):
    ax[i].axhline(0, c='red', ls='--')
    for j in range(3):
        ax[i].axvline(j, c='red', ls='--')

ax[0].scatter([0, 1, 2], [2, 0, -2], c='black', zorder=2)

fig
```

## Oppgave 2 a)

```python
x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x, y)
F = (Y**2 - X**2) * np.exp(-Y**2 - X**2)

fig, ax = plt.subplots()
ax.contour(X, Y, F, levels=50)
ax.set_xlabel('x')
ax.set_ylabel('y')
plt.show()
```

## Oppgave 2 b)

```python
fig, ax = plt.subplots(subplot_kw={'projection':'3d'}, figsize=(8, 8))
ax.plot_surface(X, Y, F, cmap='viridis')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.view_init(20, -120)
plt.show()
```

## Oppgave 3

```python
np.random.seed(1)

kurser_start = np.array([100, 100, 100])
forventning = np.array([0.0007, 0.0008, 0.0009])
usikkerhet = np.array([0.02, 0.03, 0.04])

antall_dager = 1000
kurser = np.zeros((antall_dager+1, 3))
kurser[0, :] = kurser_start

for dag in range(antall_dager):
    endring = np.random.normal(forventning, usikkerhet)
    kurser[dag+1, :] = kurser[dag, :] * (1 + endring)

fig, ax = plt.subplots(subplot_kw={'projection':'3d'}, figsize=(8, 8))
ax.plot(kurser[:, 0], kurser[:, 1], kurser[:, 2])
ax.scatter(kurser[0, 0], kurser[0, 1], kurser[0, 2], c='orange')
ax.scatter(kurser[-1, 0], kurser[-1, 1], kurser[-1, 2], c='red')
ax.set_xlabel('A'); ax.set_ylabel('B'); ax.set_zlabel('C')
ax.view_init(20, -100)
plt.show()
```
