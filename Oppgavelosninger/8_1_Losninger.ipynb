## Oppgave 1 a)

```python
import pandas as pd

inntekter = pd.Series({'Mons': 617, 'Nils': 836, 'Jens': 749, 'Hans': 583})

print(inntekter.iloc[2])
print(inntekter.iloc[[2]])

print('-' * 20)
print(inntekter.loc['Nils'])
print(inntekter.loc[['Nils']])
```

## Oppgave 1 b)

```python
print('Inntekter i stigende rekkefølge:')
print(inntekter.sort_values())

print('-' * 40)
print('Inntekter i synkende rekkefølge:')
print(inntekter.sort_values(ascending=False))

print('-' * 40)
print(f'Summen av inntektene er {inntekter.sum()},')
print(f'med et gjennomsnitt lik {inntekter.mean()}.')
```

## Oppgave 2 a)

```python
data = {'Navn': ['Kari', 'Guri', 'Siri', 'Mari'],
        'Inntekt': [851, 907, 739, 821],
        'Formue': [2984, 1183, 4092, 1625]}

df = pd.DataFrame(data)
df.index = ['Person 1', 'Person 2', 'Person 3', 'person 4']

display(df.head(3))
display(df.tail(3))
display(df.iloc[[2]])
display(df.loc[['Person 3']])
display(df.loc[df['Navn'] == 'Siri'])
```

## Oppgave 2 b)

```python
df_sortert = df.sort_values('Formue', ascending=False)
display(df_sortert)

df_filtrert = df.loc[(df['Inntekt'] > 800) & (df['Formue'] > 1500)]
display(df_filtrert)
```

## Oppgave 3

```python
resultater = {'Volum': [1120, 1360, 1510, 1470, 1440],
              'Faste': [1.27, 1.35, 1.39, 1.44, 1.61]}
df = pd.DataFrame(resultater)
df.index = [2021, 2022, 2023, 2024, 2025]

db = 3000
skatt = 0.22

df['Res før skatt'] = db * df['Volum'] - 1e6 * df['Faste']
df['Skatt'] = skatt * df['Res før skatt']
df['Res etter skatt'] = df['Res før skatt'] - df['Skatt']
df
```
