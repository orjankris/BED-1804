## Oppgave 1 a)

```python
import numpy as np

belop = 100000
rente = 0.05
aar = np.arange(11)

saldo = belop * (1 + rente) ** aar
print('Saldo hvert enkelt år:')
print(saldo.round(0))
print(f'Saldoen etter 10 år er {saldo[10]:.0f} kr')
```

## Oppgave 1 b)

```python
renter = saldo[1:] - saldo[:-1]
print('Rentebeløp hvert år:')
print(renter.round(0))

renter_okning = renter[1:] - renter[0:-1]
print('Renteøkning hvert år:')
print(renter_okning.round(0))
```

## Oppgave 2 a)

```python
alder = np.array([47, 25, 61, 34, 38, 53, 22, 59, 28, 31,
                  64, 71, 42, 50, 36, 45, 59, 24, 52, 30])

print(f'Det er {np.size(alder)} ansatte i selskapet.')
print(f'De er mellom {np.min(alder)} år og {np.max(alder)} år.')
print(f'Gjennomsnittsalderen er {np.mean(alder):.2f} år.')
```

## Oppgave 2 b)

```python
antall_u30 = np.size(alder[alder < 30])
antall_o60 = np.size(alder[alder > 60])
antall_40_50 = np.size(alder[(40 <= alder) & (alder <= 50)])

print(f'Av de ansatte er {antall_u30} under 30 år og {antall_o60} er over 60 år. ')
print(f'Det er {antall_40_50} ansatte som er mellom 40 og 50 år.')
```

## Oppgave 3
**Kommentar:** `np.random.seed(0)` gjør at randomgeneratoren starter på samme tall hver gang.  
De genererte tallene blir dermed like hver gang. Prøv gjerne også å fjerne denne.

```python
antall_forsok = 100
antall_kast_yatzy = np.zeros(antall_forsok)

np.random.seed(0)
for i in range(antall_forsok):
    kast = np.random.randint(1, 7, 5)
    antall_kast = 1
    while np.min(kast) != np.max(kast):
        kast = np.random.randint(1, 7, 5)
        antall_kast += 1
    antall_kast_yatzy[i] = antall_kast

sannsynlighet = 1 / np.mean(antall_kast_yatzy)
print(f'Anslått sannsynlighet for Yatzy er {sannsynlighet:.4%}')
```
