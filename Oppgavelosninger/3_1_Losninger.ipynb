## Oppgave 1 b)
**Kommentar:** `elif` slår ikke inn hvis `if` eller tidligere `elif` allerede har slått inn.  
Dermed vet vi at bare én brønn er lovende hvis `elif` på linje 5 slår inn.

```python
bronn_1, bronn_2 = False, False

if bronn_1 and bronn_2:
    print('Det ser lovende ut for begge brønnene')
elif bronn_1 or bronn_2:
    print('Det ser lovende ut for én av brønnene')
else:
    print('Ingen av de to brønnene ser lovende ut')
```

## Oppgave 2 a)

```python
p_1 = ['EQNR', 'DNB', 'STB', 'NHY', 'YAR']
p_2 = ['NHY', 'YAR', 'EQNR', 'SALM', 'NAS']
p_3 = ['DNB', 'STB', 'SALM', 'FRO', 'EQNR']

selskap = 'EQNR'
if selskap in p_1 and selskap in p_2 and selskap in p_3:
    print(f'Selskap {selskap} er i alle porteføljene')
```

## Oppgave 2 b)

```python
selskap = 'KOG'

if selskap not in p_1 and selskap not in p_2 and selskap not in p_3:
    print(f'Selskap {selskap} er ikke i noen av porteføljene')
if not (selskap in p_1 or selskap in p_2 or selskap in p_3):
    print(f'Selskap {selskap} er ikke i noen av porteføljene')
```

## Oppgave 3
**Kommentar:** Ved høy inntekt skal trinnskatt beregnes for flere trinn.  
Derfor må vi bruke flere `if` i stedet for `elif`, slik at alle aktuelle trinn tas med i den progressive beregningen.

```python
inntekt = 1000000
skatt = 0

if inntekt > 226100:
    skatt += 0.017 * (min(318300, inntekt) - 226100)
if inntekt > 318300:
    skatt += 0.040 * (min(725050, inntekt) - 318300)
if inntekt > 725050:
    skatt += 0.137 * (min(980100, inntekt) - 725050)
if inntekt > 980100:
    skatt += 0.168 * (min(1467200, inntekt) - 980100)
if inntekt > 1467200:
    skatt += 0.178 * (inntekt - 1467200)
                          
print(f'Trinnskatten blir {skatt:.0f} kr')
```
