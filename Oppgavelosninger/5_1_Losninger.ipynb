## Oppgave 1 a)

```python
import numpy as np

kurser_start = np.array([101.12, 39.38, 74.09, 283.24, 167.91])
kurser_slutt = np.array([109.63, 33.74, 80.61, 261.17, 188.39])

avkastninger = (kurser_slutt - kurser_start) / kurser_start
print(f'De ukentlige avkastningene er {avkastninger.round(4)}')
```

## Oppgave 1 b)

```python
kurser = np.array([[101.12, 39.38, 74.09, 283.24, 167.91], 
                   [109.63, 33.74, 80.61, 261.17, 188.39]])

avkastninger = (kurser[1] - kurser[0]) / kurser[0]
print(f'De ukentlige avkastningene er {avkastninger.round(4)}')
```

## Oppgave 2 b)

```python
avkastninger = (kurser[1:] - kurser[:-1]) / kurser[:-1]
avkastninger_log = np.log(kurser[1:] / kurser[:-1])

print(f'De vanlige avkastningene er {avkastninger.round(4)}')
print(f'De logaritmiske avkastningene er {avkastninger_log.round(4)}')
```

## Oppgave 3

```python
def resultat(inntekter, kostnader, skattesats):
    res_for_skatt = inntekter - kostnader
    skatt = skattesats * np.maximum(res_for_skatt, 0)
    res_etter_skatt = res_for_skatt - skatt
    return res_etter_skatt.round(2)

inntekter = np.array([125000, 119000, 143000])
kostnader = np.array([73000, 138000, 117000])
resultat(inntekter, kostnader, 0.22)
```
