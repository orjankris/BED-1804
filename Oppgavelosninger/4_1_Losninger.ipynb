## Oppgave 1 a)

```python
resultater = [1723926.12, 948411.38, 1413501.82, 782663.57]

resultater_tusen = []
for resultat in resultater:
    resultater_tusen.append(int(round(resultat, -3)))

resultater_tusen
```

**Kommentar:** Linje 3 - 7 kan erstattes med:

## Oppgave 1 b)

```python
maalinger = [0.4327845349, 0.5193254864, 0.4716835215, 0.4163294973, 0.4577361936]

avrunding = int(input('Ønsket antall desimaler ved avrunding: '))
maalinger_avrundet = []
for maaling in maalinger:
    maalinger_avrundet.append(round(maaling, avrunding))

maalinger_avrundet
```

## Oppgave 2 a)

```python
priser = [21, 119, 62.50, 34.90, 79, 150, 9.90, 42]

print(f'{'Antall artikler:':20}{len(priser):10}')
print(f'{'Totalsum:':20}{sum(priser):10.2f}')
print(f'{'Laveste pris:':20}{min(priser):10.2f}')
print(f'{'Høyeste pris:':20}{max(priser):10.2f}')
gjennomsnitt = sum(priser) / len(priser)
print(f'{'Gjennomsnittspris:':20}{gjennomsnitt:10.2f}')
```

## Oppgave 2 b)

```python
varer = ['Melk', 'Shampoo', 'Is', 'Brød', 'Ost', 'Grøt', 'Løk']

vare = input('Undersøk om følgende vare er i listen: ')
print(f'Antall varer er {len(varer)}, som i alfabetisk rekkefølge er gitt som:')
print(sorted(varer))
if vare in varer:
    print(f'{vare} er i listen')
else:
    print(f'{vare} er ikke i listen')
```

**Kommentar:** Linje 6 - 9 kan erstattes med:

```python
print(f'{vare} er i listen' if vare in varer else f'{vare} er ikke i listen')
```

## Oppgave 3

```python
reklame = '''BLÅBÆRHAGEN på Ås inviterer til herlige dager!
Kjøp ferske, søte bær og smak vår nydelige rømmegrøt.
Vær rask, ta turen innom og fyll opp kurven før kvelden! '''

print(reklame)
print('-' * 60)
antall_ord = len(reklame.split())
print(f'Det er {antall_ord} ord i denne reklameteksten.')

norske_tegn = []
for indeks, tegn in enumerate(reklame):
    if tegn in 'æÆøØåÅ':
        norske_tegn.append(indeks + 1)

print(f'Det er norske bokstaver på følgende posisjoner:')
print(norske_tegn)
```

**Kommentar:** Linje 10 - 16 kan erstattes med:

```python
norske_tegn = [indeks + 1 for indeks, tegn in enumerate(reklame) if tegn in "æÆøØåÅ"] 

print(f'Det er norske bokstaver på følgende posisjoner:')
print(norske_tegn)
```
